```python
import warnings
from pathlib import Path
import os 
warnings.simplefilter(action='ignore', category=FutureWarning)

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from load_data import load_all, DATA_ROOT
```

```python
triplet_df = load_all()
```

```python
triplet_df = triplet_df[triplet_df.image_set == "action"] # only need one image set since all of them were in each session
triplet_df = triplet_df[triplet_df.task == "classic"] # only need one task since both were in each session
triplet_df = triplet_df[["observer","session","date"]]
```

```python
p_obs = triplet_df[triplet_df.observer.isin(["P1", "P2"])]
o_obs = triplet_df[triplet_df.observer.isin(["O1", "O2", "O3","O4", "O5", "P1"])]
o_obs = o_obs[o_obs.session.str.contains(r"[12]")] # only include the first to sessions of P1
```

## P-Observers

```python
time_btwn_p_obs = []
for p in np.unique(p_obs.observer):
    this_p_dates = p_obs[p_obs.observer == p]
    this_p_dates = this_p_dates.sort_values("date")
    for i in range(len(this_p_dates)-1):
        this_date = this_p_dates.iloc[i].date
        next_date = this_p_dates.iloc[i+1].date
        if (next_date-this_date).round("d")>pd.Timedelta("30 days"):
            print(p)
            print(this_date, next_date)
        time_btwn_p_obs.append((next_date-this_date).round("d"))
```

```python
print("Median time between session for P observers:", np.median(time_btwn_p_obs))
print("Minimum: ", min(time_btwn_p_obs))
print("Maximum: ", max(time_btwn_p_obs))
```
